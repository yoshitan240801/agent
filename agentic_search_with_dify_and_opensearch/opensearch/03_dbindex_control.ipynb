```python
import json
import pprint

import requests
```

```python
OPENSEARCH_HOST = "https://hogehoge:10004"
OPENSEARCH_USER = "OpenSearchのユーザID"
OPENSEARCH_PASSWORD = "OpenSearchのパスワード"
```

### 登録する

##### DB indexを作成
##### https://docs.opensearch.org/latest/api-reference/index-apis/create-index/

```python
created_index_name = "my-index-name"  # 作成するindexに付ける名前
```

```python
index_mapping_payload = {"settings": {"index": {"knn": True}},
                         "mappings": {"properties": {"product_name": {"type": "text"},  # 形態素がかかりつつBM25検索
                                                     "description": {"type": "text"},  # 形態素がかかりつつBM25検索
                                                     "features": {"type": "keyword"},  # 値そのままでフィルター検索
                                                     "image_url": {"type": "keyword",  # 値そのままだけど、
                                                                   "index": False},  # 検索対象から外す
                                                     "page_url": {"type": "keyword",  # 値そのままだけど、
                                                                  "index": False},  # 検索対象から外す
                                                     "product_name_vector": {"type": "knn_vector",  # ベクトル検索
                                                                             "dimension": 1024,
                                                                             "method": {"engine": "lucene",
                                                                                        "space_type": "cosinesimil",
                                                                                        "name": "hnsw",
                                                                                        "parameters": {"ef_construction": 256,
                                                                                                       "m": 16}
                                                                                       }
                                                                            },
                                                     "description_vector": {"type": "knn_vector",  # ベクトル検索
                                                                            "dimension": 1024,
                                                                            "method": {"engine": "lucene",
                                                                                       "space_type": "cosinesimil",
                                                                                       "name": "hnsw",
                                                                                       "parameters": {"ef_construction": 256,
                                                                                                      "m": 16}
                                                                                      }
                                                                           },
                                                     "image_vector": {"type": "knn_vector",  # ベクトル検索
                                                                      "dimension": 1024,
                                                                      "method": {"engine": "lucene",
                                                                                 "space_type": "cosinesimil",
                                                                                 "name": "hnsw",
                                                                                 "parameters": {"ef_construction": 256,
                                                                                                "m": 32}
                                                                                }
                                                                     },
                                                     "features_vector": {"type": "knn_vector",  # ベクトル検索
                                                                         "dimension": 1024,
                                                                         "method": {"engine": "lucene",
                                                                                    "space_type": "cosinesimil",
                                                                                    "name": "hnsw",
                                                                                    "parameters": {"ef_construction": 256,
                                                                                                   "m": 32}
                                                                                   }
                                                                        },
                                                     "sizes": {"type": "keyword"},  # 値そのままでフィルター検索
                                                     "prices": {"type": "float"}  # 値そのままでフィルター検索
                                                    }
                                     }
                        }
```

```python
index_create_url = "{a}/{b}".format(a=OPENSEARCH_HOST, b=created_index_name)
```

```python
response = requests.put(url=index_create_url,
                        auth=(OPENSEARCH_USER,
                              OPENSEARCH_PASSWORD),
                        headers={"Content-Type": "application/json"},
                        json=index_mapping_payload,
                        verify=False
                       )
print(response.status_code)
print(response.json())
```

##### 作成済みのDB indexを確認
##### https://docs.opensearch.org/latest/api-reference/cat/cat-indices/

```python
index_check_url = "{a}/_cat/indices?format=json".format(a=OPENSEARCH_HOST)
```

```python
response_2 = requests.get(url=index_check_url,
                          auth=(OPENSEARCH_USER,
                                OPENSEARCH_PASSWORD),
                          headers={"Content-Type": "application/json"},
                          verify=False
                         )
print(response_2.status_code)
for datas in response_2.json():
    print(datas["index"])
```

### 削除する
