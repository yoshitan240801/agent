```python
import json
import os
import pprint

import dotenv
import requests
```

```python
dotenv.load_dotenv()
OPENSEARCH_HOST = "https://localhost:10004"
OPENSEARCH_USER = os.getenv("OPENSEARCH_USER")
OPENSEARCH_PASSWORD = os.getenv("OPENSEARCH_PASSWORD")
```

### 登録する

##### インデックスを作成
##### https://docs.opensearch.org/latest/api-reference/index-apis/create-index/

```python
created_index_name = "bra_panty_product_database"
```

```python
index_mapping_payload = {"settings": {"index": {"knn": True}},
                         "mappings": {"properties": {"product": {"type": "keyword",  # 値そのままで検索対象から外す
                                                                 "index": False},
                                                     "category": {"type": "keyword"},  # 値そのままでフィルター検索
                                                     "name": {"type": "text"},  # 形態素を行いつつBM25検索
                                                     "name_vector": {"type": "knn_vector",  # ベクトル検索
                                                                     "dimension": 1024,
                                                                     "method": {"engine": "lucene",
                                                                                "space_type": "cosinesimil",
                                                                                "name": "hnsw",
                                                                                "parameters": {"ef_construction": 256,
                                                                                               "m": 16}
                                                                               }
                                                                    },
                                                     "description": {"type": "text"},  # 形態素を行いつつBM25検索
                                                     "description_vector": {"type": "knn_vector",  # ベクトル検索
                                                                            "dimension": 1024,
                                                                            "method": {"engine": "lucene",
                                                                                       "space_type": "cosinesimil",
                                                                                       "name": "hnsw",
                                                                                       "parameters": {"ef_construction": 256,
                                                                                                      "m": 16}
                                                                                      }
                                                                           },
                                                     "detail": {"type": "text"},  # 形態素を行いつつBM25検索
                                                     "detail_vector": {"type": "knn_vector",  # ベクトル検索
                                                                       "dimension": 1024,
                                                                       "method": {"engine": "lucene",
                                                                                  "space_type": "cosinesimil",
                                                                                  "name": "hnsw",
                                                                                  "parameters": {"ef_construction": 256,
                                                                                                 "m": 32}
                                                                                 }
                                                                      },
                                                     "image_url": {"type": "keyword",  # 値そのままで検索対象から外す
                                                                   "index": False},
                                                     "image_vector": {"type": "knn_vector",  # ベクトル検索
                                                                      "dimension": 1024,
                                                                      "method": {"engine": "lucene",
                                                                                 "space_type": "cosinesimil",
                                                                                 "name": "hnsw",
                                                                                 "parameters": {"ef_construction": 256,
                                                                                                "m": 32}
                                                                                }
                                                                     },
                                                     "product_url": {"type": "keyword",  # 値そのままで検索対象から外す
                                                                     "index": False},
                                                     "size_price": {"type": "nested",  # sizeとpriceはネスト構造として、
                                                                    "properties": {"size": {"type": "keyword"},  # 値そのままでフィルター検索
                                                                                   "price": {"type": "float"}  # 値そのままでフィルター検索
                                                                                  }
                                                                   }
                                                    }
                                     }
                        }
```

```python
index_create_url = "{a}/{b}".format(a=OPENSEARCH_HOST, b=created_index_name)
```

```python
response = requests.put(url=index_create_url,
                        auth=(OPENSEARCH_USER,
                              OPENSEARCH_PASSWORD),
                        headers={"Content-Type": "application/json"},
                        json=index_mapping_payload,
                        verify=False
                       )
print(response.status_code)
print(response.json())
```

##### 作成済みインデックスを確認
##### https://docs.opensearch.org/latest/api-reference/cat/cat-indices/

```python
index_check_url = "{a}/_cat/indices?format=json".format(a=OPENSEARCH_HOST)
```

```python
response_2 = requests.get(url=index_check_url,
                          auth=(OPENSEARCH_USER,
                                OPENSEARCH_PASSWORD),
                          headers={"Content-Type": "application/json"},
                          verify=False
                         )
print(response_2.status_code)
for datas in response_2.json():
    print(datas["index"])
```

### 削除する

##### 対象のインデックスを削除
##### https://docs.opensearch.org/latest/api-reference/index-apis/delete-index/

```python
delete_index_name = "bra_panty_product_database"
```

```python
index_delete_url = "{a}/{b}".format(a=OPENSEARCH_HOST, b=delete_index_name)
```

```python
response_4 = requests.delete(url=index_delete_url,
                             auth=(OPENSEARCH_USER,
                                   OPENSEARCH_PASSWORD),
                             headers={"Content-Type": "application/json"},
                             verify=False
                            )
print(response_4.status_code)
print(response_4.json())
```
