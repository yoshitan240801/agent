```python
import json
import pprint

import requests
```

```python
OPENSEARCH_HOST = "https://hogehoge:10004"
OPENSEARCH_USER = "OpenSearchのユーザID"
OPENSEARCH_PASSWORD = "OpenSearchのパスワード"
```

### 検索する

##### agentic検索
##### https://docs.opensearch.org/latest/search-plugins/search-pipelines/agentic-query-translator-processor#usage

```python
OPENSEARCH_INDEX_NAME = "my-index-name"  # 03_dbindex_control.ipynbで作ったもの # 検索対象とするindexの名前
OPENSEARCH_PIPELINE_NAME = "my-pipeline-name"  # 02_agent_pipeline_control.ipynbで作ったもの # エージェント検索で用いるpipelineの名前
```

```python
agentic_search_payload = {"_source": ["product_name",
                                      "description",
                                      "features",
                                      "image_url",
                                      "page_url",
                                      "sizes",
                                      "prices"],
                          "size": 5,
                          "query": {"agentic": {"query_text": "15,000円前後で、明るめな色で鮮やかなデザインのブラジャーを探してます。Bカップです。パッドの有るものをお願いします。谷間を作れるものがイイです。"}}}
agentic_search_url = "{a}/{b}/_search?search_pipeline={c}".format(a=OPENSEARCH_HOST, b=OPENSEARCH_INDEX_NAME, c=OPENSEARCH_PIPELINE_NAME)
```

```python
response = requests.post(url=agentic_search_url,
                        auth=(OPENSEARCH_USER,
                              OPENSEARCH_PASSWORD),
                        headers={"Content-Type": "application/json"},
                        json=agentic_search_payload,
                        verify=False
                       )
print(response.status_code)
pprint.pprint(response.json())
```
